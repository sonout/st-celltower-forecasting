# Handover graph construction

This notebook uses only fictional data generated with a fixed seed. Edges represent directed successful handover counts, aggregated over the training period. Validation and test handovers are excluded. Run `python main.py --demo` first or generate the synthetic source below.


```python
from pathlib import Path
import sys
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "main.py").exists())
sys.path.insert(0, str(ROOT))
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
from src.demo import generate_synthetic_data
from src.graph import build_handover_graph
kpi_path, ho_path = generate_synthetic_data(ROOT / "data/demo/raw")
cells = pd.read_csv(kpi_path)["CELL_"].unique()
graph = build_handover_graph(pd.read_csv(ho_path), cells, start="2025-01-01", end="2025-01-05")
print(f"{graph.number_of_nodes()} fictional cells; {graph.number_of_edges()} directed handover links")
```

```python
plt.figure(figsize=(10, 6))
position = nx.circular_layout(graph)
weights = [graph[a][b]["weight"] for a, b in graph.edges()]
widths = [0.8 + 2 * weight / max(weights) for weight in weights]
nx.draw_networkx(graph, position, node_color="#dbeafe", edge_color="#64748b", node_size=900,
                 font_size=8, width=widths, arrows=True, connectionstyle="arc3,rad=0.15")
plt.title("Synthetic training-period handover graph — edge widths show counts")
plt.axis("off")
plt.show()
```
