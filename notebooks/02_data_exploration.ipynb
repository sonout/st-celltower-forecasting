# Synthetic telecom KPI exploration

The demo contains twelve fictional cells, three input KPIs, and a regular 15-minute grid. These plots illustrate the schema and daily signals; they are not evidence about a real telecommunications network.


```python
from pathlib import Path
import sys
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "main.py").exists())
sys.path.insert(0, str(ROOT))
import pandas as pd
import matplotlib.pyplot as plt
from src.demo import generate_synthetic_data
path, _ = generate_synthetic_data(ROOT / "data/demo/raw")
data = pd.read_csv(path, parse_dates=["eventdate"])
print(data.shape)
data.head()
```

```python
cell = data["CELL_"].iloc[0]
series = data[data["CELL_"] == cell].set_index("eventdate")
series[["MAX_DL_ACTIVE_UE", "DL_USER_THPUT_KBPS", "PERC_USE_PRBS_DL"]].plot(subplots=True, figsize=(12, 8))
plt.suptitle(f"Fictional cell {cell}: generated observations")
plt.tight_layout()
plt.show()
```
